```python
output_paths = {
    'train': '../../data/mixed_cv/train.tsv',
    'test': '../../data/mixed_cv/test.tsv',
    'valid': '../../data/mixed_cv/valid.tsv'
}
```

```python
def count_lines(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        return sum(1 for line in file)

print('Asimjeeg - number of samples')
for key, value in asimjeeg_paths.items():
    print(key, count_lines(value))

print('Swahili - number of samples')
for key, value in swahili_paths.items():
    print(key, count_lines(value))
```

```python
import numpy as np
import pandas as pd

ROOT = '/Users/zuzamakowska/Documents/Africa/Project/Low-resource-languages/data'

def get_clip_path(file_name, dir):
    just_file_name = os.path.split(file_name)[-1:][0]
    return os.path.join(ROOT, dir, 'clips', just_file_name)

def write_new_lines(output_path, input_path, main_ds_dir, indexes={}):
    df = pd.read_csv(input_path, sep='\t')

    if indexes:
        df = df[df.index.isin(indexes)]

    df['path'] = df['path'].apply(lambda x: get_clip_path(x, main_ds_dir))
    
    columns_to_keep = ['path', 'sentence']
    if 'eng' in df.columns:
        columns_to_keep.append('eng')
    if 'sw' in df.columns:
        columns_to_keep.append('sw')
    df = df[columns_to_keep]

    if os.path.exists(output_path):
        df.to_csv(output_path, sep='\t', index=False, mode='a', header=False)
    else:
        df.to_csv(output_path, sep='\t', index=False)

for key, value in asimjeeg_paths.items():
    write_new_lines(output_paths[key], value, '1_02_2026_tcc_cv')

for key, value in swahili_paths.items():
    no_samples = count_lines(value)
    random_indexes = np.random.choice(no_samples, size=no_samples // 4, replace=False)
    write_new_lines(output_paths[key], value, 'cv-corpus-23.0-2025-09-05/sw', set(random_indexes))            
```
